    - 'Mes de Naixement': 'Mes',

```python
df["Nom complet"] = df["Nom"] + " " + df["Cognoms"]

def nacidos_españa(pais_origen):
    if pais_origen == "Espanya":
        return True
    else:
        return False

df["Nascut Espanya"] = df["País d'origen"].apply(nacidos_españa)

df.set_index("DNI", inplace=True)

df.rename(columns={"Dia de Naixement": "Dia", "Mes de Naixement": "Mes", "Any de Naixement": "Any"}, inplace=True)

df["Gènere"] = df["Gènere"].replace({"H": "Home", "D": "Dona", "A": "Altres", "NC": "null"})

tabla_mod = df[["Nom complet","Nascut Espanya", "Dia", "Mes", "Any", "Gènere"]]
tabla_mod
```

3. Junta las columnas Fills y No Fills en una sola columna, utilizando el método .apply() y definiendo una función que resuelva el problema. La columna nueva debe llamarse "Fills" y tomar los valores "Sí" o "No".

<font color="royalblue">Objetivo:</font>  

- Juntar la información de las columnas *"Fills"* y *"No Fills"* en una solo columna. Para esto se cambia la columna "Fills" a partir de los datos de la columna "No Fills" y luego se elimina la columna "No Fills":
    - Crear la función **hijos()** con **def** para definir si una persona tiene hijos o no, usando como parámetro *hijo* en un condicional **if-else**, retornado *"No"* o *"Si"*,
    - Modificar columna *"Fills"*, a través del método **.apply()** e introduciendo como parámetro la función **hijos** sobre la columna *"No Fills"*, quedando como valores *"Si"* y *"No"*
    - Eliminar la columna *"No Fills"* con el método **.drop()**, se pasa como parámetro **columns=** la columna a borrar, junto con **inplace=True** para mantener cambio en el dataframe original *df*.

4. Crea una tabla resumen que permita ver el sueldo medio, mediano, mínimo y máximo por Género.  
Ordena la tabla en función del sueldo medio.

<font color="royalblue">Objetivo:</font>  

- Modificar la columna *"Salari mensual"* para poder aplicar funciones aritmeticas.
    - Retirar el símbolo "€" con el método **.replace()**,
    - Retirar el punto "." como separador de mil con el método **.replace()**,
    - Definir tipo de formato como *int* con el método **.astype()**
- Agrupar por la columna *"Gènere"* usando el método **.groupby()**
- Calcular las funciones pedidas a partir del método **.agg()**, como parámetro se pasa en fomato de diccionario:
    - clave: columna sobre la que se aplicaran las funciones *"Salari mensual"* y 
    - valor: las fuciones a determinar ***["mean", "median", "min", "max"]** 
- Ordenar el dataframe *"df"* con el método **.sort_values()** basados en el salario medio.

Añade las medias a los márgenes de la tabla.  
(EXTRA): Aplica formato condicional en la tabla para ver en un color más intenso los valores más elevados

<font color="royalblue">Objetivo:</font>  

- Crear la variable *tabla_resumen* con la función **pd.crosstab()** donde los parámetros son:
    - **index=**, valores de *"Génere"* como filas,
    - **columns=** valores de *"País d'origen"* como columnas,
    - **values=** valores de *"Salari mensual"* como datos para llenar el dataframe *"df"*,
    - **aggfunc=** función u operación matemática que será aplicada a los datos (values), este caso **"mean"**,
    - **margins=True** ,
- Dar formato en el display a los valores máximos de cada columna de la tabla con **.style**. Para esto se define a traves **.apply()** una función **lambda** que establece con un condicional **if-else** y un bucle **for** el fondo azul para los valores máximos *"v == x.max()"* de cada columna *x*.

```python
tabla_resumen =pd.crosstab(index=df["Gènere"], columns=df["País d'origen"], values=df["Salari mensual"], aggfunc="mean", margins=True)

tabla_resumen.style.apply(lambda x:["background: royalblue" if v == x.max() else "" for v in x])
```


6. Crea una nueva columna que sea la fecha de nacimiento en formato Datetime a partir de las columnas día, mes y año. Utilizando esta columna crea una función que dada una fecha, te calcule la edad actual a día de hoy.  
Utiliza la función que acabas de crear para generar una columna nueva en DataFrame con la edad actual.

<font color="royalblue">Objetivo:</font>  

- Importar módulo **datetime** para manejar fechas y horas,
- Crear la columna *"Data naixement"* a partir de las columnas *"Dia"*, *"Mes"* y *"Any"*, para ello armo el formato:
    - Ordenando por dia, mes y año,
    - Separando con simbolo "/" cada valor,
    - Uniendo con operador "+",
    - Modificando formato de cada columna con el método **.astype()**, de numerico a texto con parámetro *"str"*.
- Dar formato de fecha a la columna *"Data naixement"* recien creada con la función **pd.to_datetime()**. Los parámetros a usar son dato a modifica (fecha de nacimiento), **format=** tipo de formato de entrada (dia-mes-año) y **errors="coerce"** para manejar valores inválidos,
- Crear la función **edad_actual()** para calcular la edad de cada persona a la fecha actual:
    - Usa como parámetro la *"fecha_nacimiento"*,
    - Determina como variable *hoy* la fecha actual con **datetime.now()**
    - Calcula la variable *edad* restando la variable *hoy* a la fecha de nacimiento,
    - Con **.dt.days** transfoma la resta en días y luego divide entre 365 días para definir los años de edad.
- Crear columna *"Edat"* a través de la función recine creada **edad_actual()** e introduciendo como parámetro la columna *"Data naixement"*,

```python
from datetime import datetime
df["Data naixement"] = df["Dia"].astype(str) + "/" + df["Mes"].astype(str) + "/" + df["Any"].astype(str)
df["Data naixement"] = pd.to_datetime(df["Data naixement"], format="%d/%m/%Y", errors="coerce")

def edad_actual(fecha_nacimiento):
    hoy = datetime.now()
    edad = (hoy - fecha_nacimiento).dt.days // 365
    return edad
df["Edat"] = edad_actual(df["Data naixement"])
df
```

2. Utilizando un bucle, exporta en 4 archivos (formato .xlsx o .csv) los datos de cada Grupo Profesional.  
Por ejemplo: "datos_GrupA.xlsx" , "datos_GrupB.xlsx" ...  
Exporta un 5º DataFrame en formato .xlsx o .csv que contenga cuántos trabajadores hay por cada Grupo Profesional, cuál es su sueldo medio y cuál es su edad media.  

<font color="royalblue">Objetivo:</font>  

- Crear un bucle con **for** donde los 4 grupos establecen el parámetro iterador *grup* sobre la columna *"Grup Professional"*:
    - Definir la variable *df_grup* para cada almacenar los datos de cada grupo, determinado durante cada iteración de *grup*,
    - Exportar con **.to_csv()** los datos *df_grup* en el archivo *"datos_grup.csv"*, adicionalmente se elimina el espacio para cada grupo con **.replace()** para evitar espacios en el nombre del archivo. 
    - Uso del parámetro **encoding="utf-8"** para exportar de forma correcta acentos y carácteres especiales.
- Generar finalmente los archivos *"datos_GrupA.csv"*, *"datos_GrupB.csv"*, *"datos_GrupC.csv"* y *"datos_GrupD.csv"*,
- Crear un dataframe con datos de cada grupo profesional de nombre *df_Grups* con **.groupby()**, estableciendo como columna base para agrupar *"Grup Professional"* y con el método **.agg()** usando como parámetro en formato diccionario los datos a exportar:
    - "Nom":"count" número de trabajadores,
    - "Salari mensual": "mean" salario medio,
    - "Edat": "mean" edad media.
- Exportar *"df_Grups"* con el método **.to_csv()** en el archivo de nombre *"grups_professionals.csv"*.

```python
for grup in df["Grup Professional"]:
    df_grup = df[df["Grup Professional"] == grup]
    df_grup.to_csv(f"datos_{grup.replace(' ', '')}.csv", encoding="utf-8")

df_Grups=df.groupby("Grup Professional").agg({"Nom":"count", "Salari mensual": "mean", "Edat": "mean"})
df_Grups.to_csv("grups_professionals.csv")
df_Grups
```
