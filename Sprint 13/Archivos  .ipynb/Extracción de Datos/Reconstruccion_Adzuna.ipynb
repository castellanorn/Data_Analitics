### <font color="royalblue">**Reconstrucción de descripción Adzuna**</font>
Las descripciones extraidas a traves de API Adzuna estaban truncadas a 5000 caracteres.  
Por tal motivo se intenta recuperar con la url de la oferta las descripciones completas.  
- Se aplico la función *obtener_descripcion_completa(url)* al dataframe df_Adzuna.  
- Para salvaguardar las descripciones completas originales obtenidas se exportan al archivo *'df_Adzuna_v1_descripcion_completa.csv'*, esto debido a la constante actualizacion de la web de Adzuna donde entran y salen regularmente ofertas de empleo.  
- Se obtienen 51 descripciones, 25 paginas no encontradas.
- Respecto a las 25 no encontradas seran conpmetadas con las descripciones trucadas.
-  Se creo la función *filtrar_descripcion*, para extraer del Web Scraping las descripciones activas, ausentes e inactivas.
- Adicionalmente se realizo una revisión de las diferencias entre las descripciones.
- Los datos fueron exportados en archivo .csv, para posterior preprocesamiento.

```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns
import requests
from bs4 import BeautifulSoup
```

```python
# Importar los dataframes de Adzuna
df_Adzuna=pd.read_csv('df_Adzuna.csv')
df_Adzuna_desc=pd.read_csv('df_Adzuna_v1_descripcion_completa.csv')
```

```python
# Renombrar las columnas a español
df_Adzuna.rename(columns={'title': 'titulo', 'description': 'descripcion', 'location': 'ubicacion', 'company': 'empresa', 'date': 'fecha', 'url_': 'url'}, inplace=True)
```

```python
# Funcion que permite obtener la descripcion completa de una oferta laboral
def obtener_descripcion_completa(url):
    """Funcion que permite obtener la descripcion completa de una oferta laboral a partir de su url"""
    try:
        response = requests.get(url, timeout=10)
        soup = BeautifulSoup(response.text, "html.parser")

        # Así que buscamos el texto principal de la oferta
        texto = soup.get_text(separator=" ", strip=True)

        return texto
    except:
        return None

# /html/body/div[1]/main/div/section[2]/div[1]/section
```

```python
# Aplicar funcion para obtener la descripcion completa por Web Scraping
df_Adzuna["descripcion_completa"] = df_Adzuna["url"].apply(obtener_descripcion_completa)
```

```python
# Exportar el dataframe con la descripcion completa 
#df_Adzuna.to_csv("df_Adzuna_v1_descripcion_completa.csv", index=False)

# Cargar el dataframe con la descripcion completa para su posterior analisis y limpieza
df_Adzuna_desc=pd.read_csv('df_Adzuna_v1_descripcion_completa.csv')
```

```python
def duplicadas(df, columna):
    """Funcion que permite identificar las filas duplicadas en una columna específica de un DataFrame"""
    duplicadas = df[df.duplicated(subset=[columna], keep=False)]
    print(f"\n Filas duplicadas en la columna '{columna}': {len(duplicadas)}")
    return duplicadas
```

```python
duplicadas(df_Adzuna_desc, "descripcion_completa")
```

```python
def filtrar_descripcion(texto):
    """Funcion que permite filtrar la descripcion completa de una oferta laboral de Adzuna para obtener solo la parte relevante"""
    oferta_activa = "Inscribirse en esta oferta"
    oferta_ausente = "Página no encontrada"
    oferta_inactiva = "Lo sentimos, este empleo ya no está disponible"

    if oferta_activa in texto:
        descripcion = texto.split(oferta_activa)[1].strip()
        return descripcion
    elif oferta_ausente in texto:
        return None
    elif oferta_inactiva in texto:
        descripcion_no_disponible = texto.split(oferta_inactiva)[1].strip()
        descripcion_completa = descripcion_no_disponible.split("Trabajos populares")[0].strip()
        return descripcion_completa
    else:
        return texto
    
```

```python
df_Adzuna_desc["descripcion_filtrada"] = df_Adzuna_desc["descripcion_completa"].apply(filtrar_descripcion)
df_Adzuna_desc["es_truncada"] = df_Adzuna_desc["descripcion_filtrada"].isna()
df_Adzuna_desc
```

```python
df_Adzuna_desc["descripcion_final"]=df_Adzuna_desc["descripcion_filtrada"].fillna(df_Adzuna_desc["descripcion"])
df_Adzuna_desc

```

```python
duplicadas(df_Adzuna_desc, "descripcion_final")
```

```python

df_Adzuna_desc_edit = pd.merge(df_Adzuna, df_Adzuna_desc[["id", "descripcion_final", "es_truncada"]], on="id", how="left")
df_Adzuna_desc_edit
```

```python
# Exportar el DataFrame con la descripcion completa filtrada y los procesos realizados
df_Adzuna_desc.to_csv("df_Adzuna_descripcion_completa.csv", index=False)
```

```python
# Exportar el DataFrame original con la descripcion final editada 
df_Adzuna_desc_edit.to_csv("df_Adzuna_descripcion_editada.csv", index=False)
```
